This notebook contains experiments conducted during the development of my Linear Regression model. It includes different feature combinations, preprocessing approaches, and model evaluations used to understand how feature selection affected performance

Why were 5 features selected instead of 3?
During experimentation, a 3-feature model using TMAX, TMIN, and previous_day_consumption achieved a higher R² on the test set. However, I selected the 5-feature model for the final project because month_sin and month_cos capture seasonal patterns in electricity consumption and were an important part of my feature-engineering process. The goal of this project was to understand and demonstrate the complete Linear Regression workflow rather than optimize the highest possible R² score.

```python
# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

```python
#Load Dataset
df=pd.read_csv("electricity_consumption_based_weather_dataset.csv")
```

```python
#Explore Dataset
df.info()
df.columns
df.isnull().sum()
df.describe()
df["date"]=pd.to_datetime(df["date"])
df[df["AWND"].isnull()]
df['date'].is_monotonic_increasing #check for choronological ordering
```

```python
#EDA part
plt.scatter(df["TMAX"],df["daily_consumption"])#negative relationhip(x increases y decreases)
plt.scatter(df["TMIN"],df["daily_consumption"])#negative realionship(x increases y decreases)
plt.scatter(df["PRCP"],df["daily_consumption"])#no realtionship/weak realtionship why?
#bcz mostly occurences are at 0 and for higher prcp there is no data
plt.scatter(df["AWND"],df["daily_consumption"])#no realtionship/weak realtionship why?
#bcz x increases but y has no constant direction sometimes it increases and sometimes decreases
 


```

```python
#Data Cleaning
#interpolation(missing values in AWND column)
df["AWND"]=df["AWND"].interpolate()
df.isnull().sum()
#extract input and output
X=df.iloc[:,1:5]
Y=df.iloc[:,-1]
```

```python

#train test splitting 
#since it is time series data 
X_train = X.iloc[:int(len(X)*0.8)]
X_test = X.iloc[int(len(X)*0.8):]

y_train = Y.iloc[:int(len(Y)*0.8)]
y_test = Y.iloc[int(len(Y)*0.8):]
#can also use time series split built in func ,but it is mostly used for multiple train/tests validations
```

```python
#standardization
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)
```

```python
#model training
model=LinearRegression()
model.fit(X_train_scaled,y_train)
```

```python
#model testing
predictions=model.predict(X_test_scaled)
#comparison
result=pd.DataFrame({
    "Actual electricity consumption":y_test.values,
    "Predicted electricity consumption":predictions
})
result.sample()
```

```python
print("MAE",mean_absolute_error(y_test,predictions))
print("MSE",mean_squared_error(y_test,predictions))
print("RMSE",np.sqrt(mean_squared_error(y_test,predictions)))
print("R2_score",r2_score(y_test,predictions))
```

```python
#polynomial regression
from sklearn.preprocessing import PolynomialFeatures
poyl=PolynomialFeatures
for i in range(1,6):
    poly=PolynomialFeatures(degree=i)
    X_train_trans=poly.fit_transform(X_train_scaled)
    X_test_trans=poly.transform(X_test_scaled)
    lr=LinearRegression()
    lr.fit(X_train_trans,y_train)
    X_train_pred= lr.predict(X_train_trans)
    X_test_pred=lr.predict(X_test_trans)
    train_MSE=mean_squared_error(y_train,X_train_pred)
    test_MSE=mean_squared_error(y_test,X_test_pred)
    R2_score_train=r2_score(y_train,X_train_pred)
    R2_score_test=r2_score(y_test,X_test_pred)
    
    print(
            "degree :",i,
            "train_MSE:",train_MSE,
            "test_MSE:",test_MSE,
            "R2_score_train ",R2_score_train,
            "R2_score_test",R2_score_test
        )
    
    
    
    
    
```

```python
#cyclical encoding
df['month_sin']=np.sin(2*np.pi*df["month"]/12.0)
df['month_cos']=np.cos(2*np.pi*df["month"]/12.0)
df["day_num"]=df["date"].dt.dayofweek
df['day_sin']=np.sin(2*np.pi*df['day_num']/7.0)
df['day_cos']=np.cos(2*np.pi*df['day_num']/7.0)

# df=df.drop(columns=["day_of_week","month","day_num"])
X=df.iloc[:,[1,2,3,4,9,10,12,13]]
X
Y=df.iloc[:,-9]
Y
#splitting
X_train = X.iloc[:int(len(X)*0.8)]
X_test = X.iloc[int(len(X)*0.8):]

y_train = Y.iloc[:int(len(Y)*0.8)]
y_test = Y.iloc[int(len(Y)*0.8):]
#standardization
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)
#model training
lr=LinearRegression()
lr.fit(X_train_scaled,y_train)
#model prediction
predictions=lr.predict(X_test_scaled)
#comparison
result=pd.DataFrame({
    "Actual electricity consumption":y_test.values,
    "Predicted electricity consumption":predictions
})
result
```

```python
print("MAE",mean_absolute_error(y_test,predictions))
print("MSE",mean_squared_error(y_test,predictions))
print("RMSE",np.sqrt(mean_squared_error(y_test,predictions)))
print("R2_score",r2_score(y_test,predictions))
```
